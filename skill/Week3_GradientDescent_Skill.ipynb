```python
# ============================================================
# TITANIC DATASET
# LINEAR REGRESSION USING GRADIENT DESCENT
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

df = pd.read_csv("/content/Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 3. DEFINE TARGET VARIABLE
# ------------------------------------------------------------

y = df["Survived"]


# ------------------------------------------------------------
# 4. SELECT FEATURES
# ------------------------------------------------------------
# We use only numerical features.
#
# PassengerId is removed because it is only an identifier.
# Survived is removed because it is the target variable.

X = df[
    [
        "Pclass",
        "Age",
        "SibSp",
        "Parch",
        "Fare"
    ]
].copy()


# ------------------------------------------------------------
# 5. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing values before preprocessing:")
print(X.isnull().sum())


# ------------------------------------------------------------
# 6. HANDLE MISSING VALUES
# ------------------------------------------------------------
# Age contains missing values.
# Replace missing Age values with the median Age.

X["Age"] = X["Age"].fillna(X["Age"].median())


print("\nMissing values after preprocessing:")
print(X.isnull().sum())


# ------------------------------------------------------------
# 7. TRAIN-TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)


# ------------------------------------------------------------
# 8. STANDARDIZATION
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


# ------------------------------------------------------------
# 9. CONVERT TARGET TO NUMPY ARRAY
# ------------------------------------------------------------

y_train = y_train.to_numpy()
y_test = y_test.to_numpy()


# ------------------------------------------------------------
# 10. ADD BIAS / INTERCEPT COLUMN
# ------------------------------------------------------------

X_train_bias = np.c_[
    np.ones(X_train_scaled.shape[0]),
    X_train_scaled
]

X_test_bias = np.c_[
    np.ones(X_test_scaled.shape[0]),
    X_test_scaled
]


# ------------------------------------------------------------
# 11. GRADIENT DESCENT FUNCTION
# ------------------------------------------------------------

def gradient_descent(
    X,
    y,
    learning_rate=0.01,
    n_iters=1000
):

    # Number of training examples
    m = X.shape[0]

    # Number of features
    n = X.shape[1]

    # Initialize theta with zeros
    theta = np.zeros(n)

    # Store loss values
    losses = []

    # Gradient Descent iterations
    for i in range(n_iters):

        # ------------------------------------------------
        # Prediction
        # ------------------------------------------------

        y_pred = X @ theta

        # ------------------------------------------------
        # Error
        # ------------------------------------------------

        error = y_pred - y

        # ------------------------------------------------
        # Cost / Loss
        # ------------------------------------------------

        loss = (1 / (2 * m)) * np.sum(error ** 2)

        losses.append(loss)

        # ------------------------------------------------
        # Gradient
        # ------------------------------------------------

        gradient = (1 / m) * (X.T @ error)

        # ------------------------------------------------
        # Update theta
        # ------------------------------------------------

        theta = theta - learning_rate * gradient

    return theta, losses


# ------------------------------------------------------------
# 12. TRAIN MODEL
# ------------------------------------------------------------

theta, losses = gradient_descent(
    X_train_bias,
    y_train,
    learning_rate=0.01,
    n_iters=1000
)


# ------------------------------------------------------------
# 13. DISPLAY PARAMETERS
# ------------------------------------------------------------

print("\n============================================")
print("GRADIENT DESCENT PARAMETERS")
print("============================================")

print("\nIntercept:")
print(theta[0])

print("\nCoefficients:")
print(theta[1:])


# ------------------------------------------------------------
# 14. MAKE PREDICTIONS
# ------------------------------------------------------------

y_pred = X_test_bias @ theta


# ------------------------------------------------------------
# 15. EVALUATION METRICS
# ------------------------------------------------------------

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)


# ------------------------------------------------------------
# 16. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n==========================================")
print("LINEAR REGRESSION USING GRADIENT DESCENT")
print("==========================================")

print("\nMean Absolute Error (MAE):")
print(round(mae, 4))

print("\nMean Squared Error (MSE):")
print(round(mse, 4))

print("\nRoot Mean Squared Error (RMSE):")
print(round(rmse, 4))

print("\nR² Score:")
print(round(r2, 4))


# ------------------------------------------------------------
# 17. DISPLAY FIRST 10 PREDICTIONS
# ------------------------------------------------------------

print("\n==========================================")
print("ACTUAL VS PREDICTED")
print("==========================================")

results = pd.DataFrame({
    "Actual": y_test[:10],
    "Predicted": y_pred[:10]
})

print(results)


# ------------------------------------------------------------
# 18. CONVERT PREDICTIONS TO 0 OR 1
# ------------------------------------------------------------
# Since Titanic Survived is a binary target:
#
# prediction >= 0.5 -> 1 (Survived)
# prediction < 0.5  -> 0 (Did not survive)

y_pred_class = (y_pred >= 0.5).astype(int)


print("\n==========================================")
print("CLASSIFICATION PREDICTIONS")
print("==========================================")

classification_results = pd.DataFrame({
    "Actual": y_test[:10],
    "Predicted": y_pred_class[:10]
})

print(classification_results)


# ------------------------------------------------------------
# 19. CHECK FINAL LOSS
# ------------------------------------------------------------

print("\nFinal Training Loss:")
print(round(losses[-1], 6))
```
