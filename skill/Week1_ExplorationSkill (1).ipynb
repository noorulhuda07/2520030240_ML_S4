```python
# ============================================================
# TITANIC DATASET - DATA EXPLORATION
# ============================================================

# ------------------------------------------------------------
# 1. Import Libraries and Display Versions
# ------------------------------------------------------------

import sys
import pandas as pd
import numpy as np
import sklearn
import matplotlib
import seaborn as sns
import matplotlib.pyplot as plt

print("Library Versions")
print("-" * 40)
print(f"Pandas         : {pd.__version__}")
print(f"NumPy          : {np.__version__}")
print(f"Scikit-learn   : {sklearn.__version__}")
print(f"Matplotlib     : {matplotlib.__version__}")
print(f"Seaborn        : {sns.__version__}")


# ------------------------------------------------------------
# 2. Load Titanic Dataset
# ------------------------------------------------------------

from pandas import read_csv

df = read_csv("/content/Titanic-Dataset.csv")


# ------------------------------------------------------------
# 3. Display Last 5 Records
# ------------------------------------------------------------

# Displays the last five records of the dataset
df.tail(5)


# ------------------------------------------------------------
# 4. Display PassengerId and Shape
# ------------------------------------------------------------

df['PassengerId']

df.shape


# ------------------------------------------------------------
# 5. Display Name Column
# ------------------------------------------------------------

df['Name']


# ------------------------------------------------------------
# 6. Display Data Types
# ------------------------------------------------------------

df.dtypes


# ------------------------------------------------------------
# 7. Count Number of Columns by Data Type
# ------------------------------------------------------------

df.dtypes.value_counts()


# ------------------------------------------------------------
# 8. Check Missing Values
# ------------------------------------------------------------

missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_data


# ------------------------------------------------------------
# 9. Visualize Missing Values using Heatmap
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

# cbar=False hides the color bar
# yticklabels=False hides row numbers for cleaner visualization
sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("missing")
plt.show()
```
