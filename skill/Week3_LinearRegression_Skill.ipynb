```python
# ============================================================
# TITANIC DATASET - LINEAR REGRESSION
# ============================================================

# 1. Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 2. Load Dataset
# ============================================================

df = pd.read_csv("/content/Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 3. Display First Few Rows
# ============================================================

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 4. Select Target Variable and Features
# ============================================================

# Target variable
# We are predicting Fare
y = df["Fare"]

# Input features
features = [
    "Age",
    "Pclass",
    "Sex",
    "SibSp",
    "Parch"
]

X = df[features].copy()


# ============================================================
# 5. Define Numerical and Categorical Features
# ============================================================

numerical_features = [
    "Age",
    "SibSp",
    "Parch"
]

categorical_features = [
    "Sex",
    "Pclass"
]


# ============================================================
# 6. Handle Missing Values
# ============================================================

# Fill missing numerical values with median
for col in numerical_features:
    if col in X.columns:
        X[col] = X[col].fillna(X[col].median())


# Fill missing categorical values with mode
for col in categorical_features:
    if col in X.columns:
        X[col] = X[col].fillna(X[col].mode()[0])


# Check for missing values
print("\nMissing Values After Imputation:")
print(X.isnull().sum())


# ============================================================
# 7. Split Dataset into Training and Testing Sets
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)


# ============================================================
# 8. Create Preprocessor
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[

        # Numerical features
        (
            "num",
            StandardScaler(),
            numerical_features
        ),

        # Categorical features
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_features
        )
    ]
)


# ============================================================
# 9. Create Linear Regression Pipeline
# ============================================================

linear_model = Pipeline(
    steps=[

        ("preprocessor", preprocessor),

        (
            "regressor",
            LinearRegression()
        )
    ]
)


# ============================================================
# 10. Train the Model
# ============================================================

linear_model.fit(
    X_train,
    y_train
)

print("\nModel Training Completed Successfully!")


# ============================================================
# 11. Make Predictions
# ============================================================

y_pred = linear_model.predict(X_test)


# ============================================================
# 12. Calculate Evaluation Metrics
# ============================================================

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)


# ============================================================
# 13. Display Results
# ============================================================

print("\n============================================")
print("LINEAR REGRESSION RESULTS")
print("============================================")

print("\nMean Absolute Error (MAE):")
print(round(mae, 4))

print("\nMean Squared Error (MSE):")
print(round(mse, 4))

print("\nRoot Mean Squared Error (RMSE):")
print(round(rmse, 4))

print("\nR² Score:")
print(round(r2, 4))


# ============================================================
# 14. Display Actual vs Predicted Values
# ============================================================

results = pd.DataFrame({
    "Actual Fare": y_test.values,
    "Predicted Fare": y_pred
})

print("\n============================================")
print("ACTUAL VS PREDICTED FARE")
print("============================================")

print(results.head(10))


# ============================================================
# 15. Predict Fare for a New Passenger
# ============================================================

new_passenger = pd.DataFrame({
    "Age": [25],
    "Pclass": [3],
    "Sex": ["male"],
    "SibSp": [0],
    "Parch": [0]
})

predicted_fare = linear_model.predict(new_passenger)

print("\n============================================")
print("NEW PASSENGER PREDICTION")
print("============================================")

print("Passenger Details:")
print(new_passenger)

print("\nPredicted Fare:")
print(round(predicted_fare[0], 2))
```
